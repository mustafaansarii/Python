```python
from pytube import Playlist
import re

def sanitize_filename(title):
    # Remove invalid characters from the filename
    return re.sub(r'[<>:"/\\|?*]', '', title)

def download_playlist(playlist_url, resolution='1080p'):
    # Create a Playlist object
    playlist = Playlist(playlist_url)
    
    # Print the number of videos in the playlist
    print(f'Downloading {len(playlist.video_urls)} videos from playlist: {playlist.title}')

    for video_url in playlist.video_urls:
        video = playlist.get(video_url)
        
        # Get the stream with the desired resolution
        stream = video.streams.filter(res=resolution, file_extension='mp4').first()

        if stream:
            title = f'DR NO {video.title}'  # Prepend "DR NO" to the video title
            sanitized_title = sanitize_filename(title)  # Sanitize the filename
            print(f'Downloading: {sanitized_title}')
            stream.download(filename=sanitized_title)  # Download with the new filename
        else:
            print(f'No stream found for resolution {resolution} in video: {video.title}')

if __name__ == "__main__":
    playlist_url = 'https://youtube.com/playlist?list=PLA3GkZPtsafacdBLdd3p1DyRd5FGfr3Ue'
    download_playlist(playlist_url)
```

```python
def remove_duplicates(arr):
    if not arr:
        return 0

    unique_position = 0
    
    for i in range(1, len(arr)):
        if arr[unique_position] != arr[i]:
            unique_position += 1
            arr[unique_position] = arr[i]
    
    return unique_position + 1

```

```python
arr = [2, 2, 2, 2, 2]
length = remove_duplicates(arr)
print(f"Modified array: {arr[:length]}")
print(f"Length of modified array: {length}")
```
